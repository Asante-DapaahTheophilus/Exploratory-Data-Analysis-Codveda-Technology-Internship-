**Distribution of Sales Amount, I wanted to drive insights on customer spending behaviour and also check if there are some anomalities in sales**

```python
df.hist(["Sales_Amount"], bins = 30)
plt.title("Distribution of Sales Amount")
plt.xlabel("Sales_Amount")
plt.ylabel("Frequency")
plt.figure(figsize=(9, 5))
plt.show()
```

**Distribution of Quantity Sold (purchasing behavior of customer and quantity of orders )**

```python
df.hist(["Quantity_Sold"], bins=20)
plt.title("Distribution of Quantity Sold")
plt.xlabel("Quantity_Sold")
plt.ylabel("Frequency")
plt.figure(figsize=(9, 5))
plt.show()
```

**Relatiosnhip between price and sales amount, let's find out if pricing affect revenue**

```python
df.plot(kind="scatter", x="Unit_Price", y="Sales_Amount")
plt.title("Relationship between Price and Sales_Amount")
plt.xlabel("Unit_Price")
plt.ylabel("Sales_Amount")
plt.figure(figsize=(9, 5))
plt.show
```

**Relationship between quantity sold and revenue, let's see if more purchases significantly increase revenue**

```python
plt.figure(figsize=(9, 5))
df.plot(kind="scatter", x="Quantity_Sold", y="Sales_Amount")
plt.title("Relationship between Quantity Sold and Sales_Amount")
plt.xlabel("Quantity_Sold")
plt.ylabel("Sales_Amount")
```

**Box plot of Sales_Amount, let's find out if there are some outliers in the sales_amount or potential fraud**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Sales_Amount"])
plt.title("Box Plot of Sales_Amount")
plt.xlabel("Sales_Amount")
plt.show()
```

**Let's find out if there are some unsual prices or outliers in the of pricing of goods**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Unit_Price"])
plt.title("Box Plot of Unit_Price")
plt.xlabel("Unit_Price")
plt.show()
```

**Let's see the best performing regions**

```python
region_sales = df.groupby("Region")["Sales_Amount"].sum()
region_sales.plot(kind="bar", figsize=(8,5))
plt.title("Total Sales Amount by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

**Now let's find out the best selling product**

```python
region_sales = df.groupby("Product_Category")["Sales_Amount"].sum()
region_sales.plot(kind="bar", figsize=(9,5))
plt.title("Total Sales by Product_Category")
plt.xlabel("Product_Category")
plt.ylabel("Total Sales")
plt.show()
```

**Let's find out what channel of purchase most customers are convinient with**

```python
sns.countplot(x="Sales_Channel", data=df)
plt.title("Channel of Purchase")
plt.xlabel("Sales_Channel")
plt.ylabel("Count")
plt.show()
```

**Monthly sales over time**

```python
df["Sale_Date"]=pd.to_datetime(df["Sale_Date"])
df
```

```python
monthly_sales = df.groupby(df["Sale_Date"].dt.to_period("M"))["Sales_Amount"].sum()
monthly_sales.plot(kind="line", figsize=(7,5))
plt.title("Monthly Sales")
```

**Let's dive into the relatiosnhip between Numerical columns**

```python
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

**Sales by Customer type, Let's see customer that brought in most revenue**

```python
customer_type = df.groupby("Customer_Type")["Sales_Amount"].sum()
customer_type.plot(kind="bar", figsize=(8,5))
plt.title("Sales by Customer Type")
plt.xlabel("Customer_Type")
plt.ylabel("Total Sales")
plt.show()
```

**Top performing employees**

```python
top_reps=df.groupby("Sales_Rep")["Sales_Amount"].sum().sort_values(ascending=False).head()
top_reps.plot(kind="bar", figsize=(8,5))
plt.title("Top 5 Performing Sales Representatives")
plt.xlabel("Top representatives")
plt.ylabel("Total Sales")
plt.show()
```
